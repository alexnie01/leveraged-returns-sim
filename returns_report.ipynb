# Returns Report

Single notebook for cumulative-return analysis:

1. **Single path** — one baseline series and a 3× daily-returns variant
2. **Monte Carlo** — many simulated trials with mean vs median comparison

Configure inputs below, run all cells, and share the exported HTML report.

```python
from pathlib import Path

import numpy as np

from returns_report import (
    ReportConfig,
    build_combined_figure,
    build_report,
    export_report,
)

# --- Inputs ---
USE_HISTORICAL = False  # True: download ticker data and calibrate drift/vol from sample

# Used when USE_HISTORICAL is False
ANNUAL_DRIFT = 0.08
ANNUAL_VOLATILITY = 0.20

# Used when USE_HISTORICAL is True
TICKER = "SPY"

# Simulation settings
N_TRIALS = 1000
N_DAYS = 1000
RANDOM_SEED = 42
SHOW_MC_TRIALS = True
HTML_OUTPUT = Path("returns_report.html")

config = ReportConfig(
    use_historical=USE_HISTORICAL,
    ticker=TICKER,
    annual_drift=ANNUAL_DRIFT,
    annual_volatility=ANNUAL_VOLATILITY,
    n_trials=N_TRIALS,
    n_days=N_DAYS,
    random_seed=RANDOM_SEED,
    show_mc_trials=SHOW_MC_TRIALS,
    html_output=HTML_OUTPUT,
)

rng = np.random.default_rng(RANDOM_SEED)
data = build_report(config, rng=rng)

print("Parameters used for Monte Carlo:")
print(f"  Annual drift:      {data.annual_drift:+.2%}")
print(f"  Annual volatility: {data.annual_volatility:.2%}")
print(f"  Single path:       {data.single_path.source.value}")
print(f"  Trials × days:     {config.n_trials:,} × {config.n_days}")
print("\nFinal Monte Carlo cumulative return:")
for label, value in data.mc_summary.items():
    print(f"  {label.replace('_', ' ')}: {value:+.2%}")
```

```python
fig = build_combined_figure(data, show_mc_trials=config.show_mc_trials)
fig.show()
```

```python
output_path = export_report(fig, config.html_output)
print(f"Report saved to: {output_path.resolve()}")
```
